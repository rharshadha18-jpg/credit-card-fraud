```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

os.makedirs('../reports/figures', exist_ok=True)
df = pd.read_csv('../data/creditcard.csv')
print(df.shape)
df.head()
```

```python
print('Missing values:', df.isnull().sum().sum())
print('Duplicates:', df.duplicated().sum())
print(df['Class'].value_counts())
print(f"Fraud ratio: {df['Class'].mean()*100:.3f}%")
```

```python
# Class distribution
ax = df['Class'].value_counts().plot(kind='bar', color=['steelblue','tomato'])
ax.set_xticklabels(['Genuine (0)', 'Fraud (1)'], rotation=0)
ax.set_title('Class Distribution')
ax.set_ylabel('Count')
plt.tight_layout()
plt.savefig('../reports/figures/class_distribution.png', dpi=150)
plt.show()
```

```python
# Amount distribution: fraud vs genuine
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
for ax, col in zip(axes, ['Amount', 'Time']):
    df[df['Class']==0][col].plot(kind='hist', bins=50, alpha=0.6, label='Genuine', ax=ax)
    df[df['Class']==1][col].plot(kind='hist', bins=50, alpha=0.6, label='Fraud', ax=ax)
    ax.set_title(col)
    ax.legend()
plt.tight_layout()
plt.savefig('../reports/figures/amount_time_dist.png', dpi=150)
plt.show()
```

```python
# Top feature correlations with Class
corr = df.corr()['Class'].drop('Class').abs().sort_values(ascending=False)
corr.head(15).plot(kind='bar', figsize=(10, 4), title='Feature Correlation with Class (abs)')
plt.tight_layout()
plt.savefig('../reports/figures/feature_correlation.png', dpi=150)
plt.show()
```
